## Creación de tablas:

En el siguiente notebook se crean las 5 tablas correspondientes al proyecto de apuestas en linea, las tablas se crean en base al modelo relaciónal del entregable 1.

```sql
USE CATALOG workspace;
USE SCHEMA apuestas_db;
```

```sql
-- 1. Estado apuesta
CREATE TABLE IF NOT EXISTS estado_apuesta (
    id_estado INT NOT NULL,
    nombre_estado STRING NOT NULL,
    CONSTRAINT pk_estado PRIMARY KEY (id_estado) NOT ENFORCED
);
```

```sql
-- 2. Usuario
CREATE TABLE IF NOT EXISTS usuario (
    id_usuario BIGINT GENERATED ALWAYS AS IDENTITY NOT NULL,
    nombre STRING NOT NULL,
    apellido STRING NOT NULL,
    numero_documento STRING NOT NULL,
    tipo_documento STRING NOT NULL,
    fecha_registro TIMESTAMP,
    saldo DECIMAL(18,2),
    CONSTRAINT pk_usuario PRIMARY KEY (id_usuario) NOT ENFORCED
);
```

```sql
-- 3. Evento
CREATE TABLE IF NOT EXISTS evento (
    id_evento BIGINT GENERATED ALWAYS AS IDENTITY NOT NULL,
    tipo_deporte STRING NOT NULL,
    participantes STRING NOT NULL,
    fecha_evento TIMESTAMP,
    resultado STRING,
    CONSTRAINT pk_evento PRIMARY KEY (id_evento) NOT ENFORCED
);
```

```sql
-- 4. Apuesta
CREATE TABLE IF NOT EXISTS apuesta (
    id_apuesta BIGINT GENERATED ALWAYS AS IDENTITY NOT NULL,
    id_usuario BIGINT,
    id_evento BIGINT,
    monto DECIMAL(18,2) NOT NULL,
    cuota DECIMAL(10,2) NOT NULL,
    fecha_apuesta TIMESTAMP,
    id_estado_apuesta INT,
    CONSTRAINT pk_apuesta PRIMARY KEY (id_apuesta) NOT ENFORCED,
    CONSTRAINT fk_apuesta_usuario FOREIGN KEY (id_usuario) REFERENCES usuario(id_usuario) NOT ENFORCED,
    CONSTRAINT fk_apuesta_evento FOREIGN KEY (id_evento) REFERENCES evento(id_evento) NOT ENFORCED,
    CONSTRAINT fk_apuesta_estado FOREIGN KEY (id_estado_apuesta) REFERENCES estado_apuesta(id_estado) NOT ENFORCED
);
```

```sql
-- Ajuste: agregar fecha a la apuesta 
ALTER TABLE apuesta ADD COLUMN fecha_apuesta TIMESTAMP;
```

```sql
-- Vaciamos las filas viejas/incompletas, mantenemos el schema ya corregido
TRUNCATE TABLE apuesta;
TRUNCATE TABLE transaccion;
```

```sql
-- 5. Transaccion
CREATE TABLE IF NOT EXISTS transaccion (
    id_transaccion BIGINT GENERATED ALWAYS AS IDENTITY NOT NULL,
    id_usuario BIGINT,
    id_apuesta BIGINT,
    monto DECIMAL(18,2) NOT NULL,
    tipo_transaccion STRING NOT NULL,
    fecha_transaccion TIMESTAMP,
    CONSTRAINT pk_transaccion PRIMARY KEY (id_transaccion) NOT ENFORCED,
    CONSTRAINT fk_transaccion_usuario FOREIGN KEY (id_usuario) REFERENCES usuario(id_usuario) NOT ENFORCED,
    CONSTRAINT fk_transaccion_apuesta FOREIGN KEY (id_apuesta) REFERENCES apuesta(id_apuesta) NOT ENFORCED
);
```
